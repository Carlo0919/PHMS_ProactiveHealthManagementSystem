```python
# Data Preprocessing
# Load the dataset
import pandas as pd

file_path = r"C:\Users\khiew\Downloads\DiseaseSymptom_OriginalDataset.csv"

import pandas as pd
data = pd.read_csv(file_path)

original_row_count = data.shape[0]
print(f"Original number of rows: {original_row_count}")

# Detect and Remove Duplicate Rows
data_cleaned = data.drop_duplicates()

# Check the Shape of the Dataset after Removing Duplicates
cleaned_row_count = data_cleaned.shape[0]
print(f"Number of rows after removing duplicates: {cleaned_row_count}")

# Calculate the Number of Rows Eliminated
rows_eliminated = original_row_count - cleaned_row_count
print(f"Number of rows eliminated: {rows_eliminated}")
```
